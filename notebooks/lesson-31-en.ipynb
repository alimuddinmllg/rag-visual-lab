```python
# NalarNusa RAG Visual Lab · Lesson 6 · A RAG pipeline from scratch
# Only Python + numpy. Each stage below is the code the visual runs.
import math
import re
import numpy as np

def parse_document(text):
    """Split the document into words and remember which section each word is in."""
    sections, words = [], []
    for line in text.split("\n"):
        t = line.strip()
        if not t:
            continue
        if t.startswith("## "):
            sections.append(t[3:])
            continue
        for w in t.split():
            words.append((w, max(0, len(sections) - 1)))
    if not sections:
        sections.append("Document")
    return sections, words

def chunk_document(words, size, overlap):
    """Cut the word list into chunks of `size` words that overlap by `overlap` words."""
    step = max(5, size - overlap)
    chunks = []
    start = 0
    while start < len(words):
        ws = words[start:start + size]
        counts = {}
        for _, sec in ws:
            counts[sec] = counts.get(sec, 0) + 1
        ranked = sorted(counts.items(), key=lambda kv: (-kv[1], kv[0]))
        mixed = len(ranked) > 1 and ranked[0][1] / len(ws) < 0.75
        chunks.append({
            "id": len(chunks),
            "text": " ".join(w for w, _ in ws),
            "sec": ranked[0][0],
            "sec2": ranked[1][0] if mixed else None,
        })
        if start + size >= len(words):
            break
        start += step
    return chunks

def stem(w, lang):
    """A tiny English stemmer: printing -> print, prints -> print."""
    if lang != "en":
        return w
    if len(w) > 5 and w.endswith("ing"):
        return w[:-3]
    if len(w) > 4 and w.endswith("ies"):
        return w[:-3] + "y"
    if len(w) > 3 and w.endswith("s") and not w.endswith("ss"):
        return w[:-1]
    return w

def tokenize(text, stop, lang):
    words = re.findall(r"[a-z0-9]+", text.lower())
    return [stem(t, lang) for t in words if len(t) > 1 and t not in stop]

def vectorize(tokens, idf):
    """TF-IDF vector, scaled to length 1. Stored as {word: weight}."""
    tf = {}
    for t in tokens:
        if t in idf:
            tf[t] = tf.get(t, 0) + 1
    v = {t: (1 + math.log(c)) * idf[t] for t, c in tf.items()}
    norm = math.sqrt(sum(w * w for w in v.values())) or 1
    return {t: w / norm for t, w in v.items()}

def embed_chunks(chunks, stop, lang):
    toks = [tokenize(c["text"], stop, lang) for c in chunks]
    df = {}
    for ts in toks:
        for t in dict.fromkeys(ts):          # each word once per chunk
            df[t] = df.get(t, 0) + 1
    n = len(chunks)
    idf = {t: math.log((n + 1) / (d + 1)) + 1 for t, d in df.items()}
    vocab = list(idf.keys())
    vectors = [vectorize(ts, idf) for ts in toks]
    return idf, vocab, vectors

def pca_2d(vectors, vocab):
    """Squash the vectors to 2D for the map (PCA by power iteration)."""
    index = {t: i for i, t in enumerate(vocab)}
    X = np.zeros((len(vectors), len(vocab)))
    for i, v in enumerate(vectors):
        for t, w in v.items():
            X[i, index[t]] = w
    mean = X.mean(axis=0)
    Xc = X - mean
    K = Xc @ Xc.T
    comps = []
    for c in range(2):
        v = np.array([1 + ((i * 7 + c * 3) % 5) * 0.13 for i in range(len(vectors))])
        lam = 0.0
        for _ in range(150):
            w = K @ v
            nm = np.linalg.norm(w) or 1
            v, lam = w / nm, nm
        if v[np.argmax(np.abs(v))] < 0:
            v = -v
        comps.append((v, lam))
        K = K - lam * np.outer(v, v)
    coords = np.stack([v * math.sqrt(max(lam, 1e-12)) for v, lam in comps], axis=1)
    axes = [Xc.T @ (v / math.sqrt(max(lam, 1e-12))) for v, lam in comps]
    return coords, mean, axes, index

def embed_question(question, idf, stop, lang):
    q_tokens = tokenize(question, stop, lang)
    qv = vectorize(q_tokens, idf)
    known = list(dict.fromkeys(t for t in q_tokens if t in idf))
    return qv, known

def cosine(a, b):
    """Both vectors have length 1, so cosine similarity is just the dot product."""
    if len(a) > len(b):
        a, b = b, a
    return sum(w * b[t] for t, w in a.items() if t in b)

def retrieve(qv, vectors, top_k):
    scores = [cosine(qv, v) if qv else 0.0 for v in vectors]
    order = sorted(range(len(scores)), key=lambda i: -scores[i])
    return scores, [(i, scores[i]) for i in order[:top_k]]

PROMPT = {
    "en": dict(
        system="You are a helpful assistant for the NusaBot Robotics Lab.\n"
               "Answer ONLY using the context below and cite sources like [1].\n"
               "If the answer is not in the context, say you don't know.",
        context="Context:", none="(no relevant chunks found)", question="Question:", answer="Answer:",
        refuse="I don't know. The document doesn't contain anything about this question."),
    "id": dict(
        system="Anda adalah asisten untuk NusaBot Robotics Lab.\n"
               "Jawab HANYA berdasarkan konteks di bawah dan sebutkan sumber seperti [1].\n"
               "Jika jawabannya tidak ada di konteks, katakan Anda tidak tahu.",
        context="Konteks:", none="(tidak ada potongan yang relevan)", question="Pertanyaan:", answer="Jawaban:",
        refuse="Saya tidak tahu. Dokumen tidak berisi informasi tentang pertanyaan ini."),
}

def build_prompt(question, used, chunks, sections, lang):
    T = PROMPT.get(lang, PROMPT["en"])
    ctx = "\n\n".join(f"[{r + 1}] ({sections[chunks[i]['sec']]}) {chunks[i]['text']}"
                      for r, (i, _) in enumerate(used))
    return f"{T['system']}\n\n{T['context']}\n{ctx or T['none']}\n\n{T['question']} {question}\n{T['answer']}"

def extract_answer(used, chunks, known, idf, stop, lang):
    """Stand-in for an LLM: pick the best-matching sentences and cite them."""
    T = PROMPT.get(lang, PROMPT["en"])
    if not used:
        return T["refuse"]
    qset = set(known)
    cands = []
    for r, (i, _) in enumerate(used):
        for sent in re.split(r"(?<=[.!?])\s+", chunks[i]["text"]):
            sc = sum(idf.get(x, 0) for x in dict.fromkeys(tokenize(sent, stop, lang)) if x in qset)
            if sc > 0:
                cands.append((sent, sc, r))
    cands.sort(key=lambda c: -c[1])
    picks = []
    for sent, sc, r in cands:
        if len(picks) >= 2:
            break
        if len(re.split(r"\s+", sent)) < 5:
            continue
        if any(sent[:40] in p[0] or p[0][:40] in sent for p in picks):   # overlap repeats text
            continue
        if picks and sc < picks[0][1] * 0.6:
            break
        picks.append((sent, sc, r))
    if not picks:
        return T["refuse"]
    fmt = lambda s: ("" if re.match(r"[A-Z0-9]", s) else "…") + s + ("" if re.search(r"[.!?]$", s) else "…")
    return " ".join(f"{fmt(s)} [{r + 1}]" for s, _, r in picks)

def run_pipeline(document, question, size, overlap, top_k, min_sim, lang, stop, verbose=False):
    log = print if verbose else (lambda *a: None)
    sections, words = parse_document(document)
    chunks = chunk_document(words, size, overlap)
    log(f"chunk    size={size} overlap={overlap} -> {len(chunks)} chunks")

    idf, vocab, vectors = embed_chunks(chunks, stop, lang)
    coords, mean, axes, _ = pca_2d(vectors, vocab)
    log(f"embed    TF-IDF, {len(vocab)} terms -> {len(vectors)} vectors")

    qv, known = embed_question(question, idf, stop, lang)
    log(f"query    known words: {', '.join(known) or '(none)'}")

    scores, top = retrieve(qv, vectors, top_k)
    used = [(i, s) for i, s in top if s >= min_sim and known]
    log("retrieve " + ", ".join(f"#{i + 1} {s:.2f}" for i, s in top))

    prompt = build_prompt(question, used, chunks, sections, lang)
    answer = extract_answer(used, chunks, known, idf, stop, lang)
    log(f"generate {len(used)} chunks in prompt\n\nAnswer: {answer}")

    query_xy = None
    if known:
        q = np.zeros(len(vocab))
        for t, w in qv.items():
            q[vocab.index(t)] = w
        query_xy = [float((q - mean) @ a) for a in axes]
    return {
        "n_chunks": len(chunks),
        "vocab_size": len(vocab),
        "known": known,
        "top": [[i, s] for i, s in top],
        "used": [i for i, _ in used],
        "coords": coords.tolist(),
        "query_xy": query_xy,
        "prompt": prompt,
        "answer": answer,
    }
```

```python
# NalarNusa RAG Visual Lab · Lesson 31 · Monitoring in production: a 30-day query log, daily metrics, alerts, drill-down
# The log is simulated (seeded, so it is the same every run), but every query really goes through retrieval.
import math
import re

def rng(seed):
    s = seed % 2147483646 + 1
    def rand():
        nonlocal s
        s = s * 16807 % 2147483647
        return s / 2147483647
    return rand

def without_section(document, title):
    """The incident: a sync bug drops one page from the index (lesson 26)."""
    return "\n".join(p for p in re.split(r"\n(?=## )", document) if not p.startswith("## " + title))

def make_index(document, stop, lang):
    _, words = parse_document(document)
    chunks = chunk_document(words, 40, 8)
    idf, _, vectors = embed_chunks(chunks, stop, lang)
    return chunks, idf, vectors

def coverage(question, index, stop, lang):
    """Share of the question's words found in the best chunk. Under half = an empty retrieval."""
    chunks, idf, vectors = index
    words = list(dict.fromkeys(tokenize(question, stop, lang)))
    qv = vectorize(words, idf)
    scores = [cosine(qv, v) if qv else 0.0 for v in vectors]
    best = min(range(len(scores)), key=lambda i: (-scores[i], i))
    have = set(tokenize(chunks[best]["text"], stop, lang))
    return sum(t in have for t in words) / max(1, len(words))

def simulate(document, pools, broken_title, stop, lang, days=30, seed=11, min_cover=0.5):
    """Each day: a number of queries, each with a retrieval score, a latency and maybe a thumbs up/down.
    Planted problems: a slow model provider on days 6-8, the 3D printing page missing on days 13-16,
    and from day 22 users asking about a new machine the documents don't cover."""
    r = rng(seed)
    indexes = {"ok": make_index(document, stop, lang), "broken": make_index(without_section(document, broken_title), stop, lang)}
    cache, log = {}, []
    answerable = pools["answerable"] + pools["printing"]               # printing questions are asked twice as often
    for day in range(1, days + 1):
        version = "broken" if 13 <= day <= 16 else "ok"
        weekend = day % 7 in (6, 0)
        n = math.floor(40 * (0.6 if weekend else 1) * (1 + 0.01 * day) + 0.5) + math.floor(r() * 8)
        for _ in range(n):
            u, new_share = r(), (0.18 if day >= 22 else 0.0)
            if u < new_share:
                pool = pools["new"]
            elif u < new_share + 0.06:
                pool = pools["unanswerable"]
            else:
                pool = answerable
            q = pool[math.floor(r() * len(pool))]
            if (version, q) not in cache:
                cache[(version, q)] = coverage(q, indexes[version], stop, lang)
            cover = cache[(version, q)]
            empty = cover < min_cover
            ms = (700 + r() * 600) * (2.5 if 6 <= day <= 8 else 1)
            gives, down = r() < 0.3, r() < (0.8 if empty else 0.08)
            log.append({"day": day, "q": q, "cover": cover, "empty": empty, "ms": ms, "feedback": ("down" if down else "up") if gives else None})
    return log

def daily(log, days=30):
    out = []
    for d in range(1, days + 1):
        rows = [x for x in log if x["day"] == d]
        fb = [x for x in rows if x["feedback"]]
        ms = sorted(x["ms"] for x in rows)
        out.append({"day": d, "volume": len(rows), "empty": sum(x["empty"] for x in rows) / len(rows),
                    "down": sum(x["feedback"] == "down" for x in fb) / len(fb) if fb else 0.0,
                    "p95": ms[math.ceil(0.95 * len(ms)) - 1]})
    return out

def alerts(days, rules, window):
    """A rule fires when its metric is over the threshold for `window` days in a row; one alert per episode."""
    fired = []
    for metric, limit in rules.items():
        run, on = 0, False
        for d in days:
            run = run + 1 if d[metric] > limit else 0
            if run >= window and not on:
                fired.append({"metric": metric, "day": d["day"]})
            on = run >= window
    return sorted(fired, key=lambda a: (a["day"], a["metric"]))

def drill(log, day, top=5):
    """What did people ask, and not get, on the alert day?"""
    counts = {}
    for x in log:
        if x["day"] == day and x["empty"]:
            counts[x["q"]] = counts.get(x["q"], 0) + 1
    return sorted(counts.items(), key=lambda kv: (-kv[1], kv[0]))[:top]

def demo(document, pools, broken_title, rules, window, lang, stop):
    stop = set(stop)
    log = simulate(document, pools, broken_title, stop, lang)
    days = daily(log)
    fired = alerts(days, rules, window)
    empty_alerts = [a["day"] for a in fired if a["metric"] == "empty"]
    drilled = {d: drill(log, d) for d in empty_alerts}
    print(f"{len(log)} queries over {len(days)} days")
    for a in fired:
        print(f"day {a['day']:2d}: ALERT {a['metric']}")
    for d, rows in drilled.items():
        print(f"day {d} empty retrievals:", rows)
    return {"n": len(log), "days": days, "alerts": fired, "drill": {str(d): rows for d, rows in drilled.items()}}
```

```python
# NalarNusa RAG Visual Lab · Lesson 31 · Production logging and alerting with JSON lines and pandas
# pip install pandas
import json
import time
import pandas as pd

LOG = "rag_queries.jsonl"


def log_query(question, chunks_used, best_cover, ms, answer):
    """Call this once per answered question. One JSON object per line is easy to append and to load."""
    with open(LOG, "a", encoding="utf-8") as f:
        f.write(json.dumps({"ts": time.time(), "q": question, "chunks": chunks_used, "cover": best_cover,
                            "empty": best_cover < 0.5, "ms": ms, "answer": answer, "feedback": None}) + "\n")


def daily(path=LOG):
    df = pd.read_json(path, lines=True)
    df["day"] = pd.to_datetime(df["ts"], unit="s").dt.date
    return df.groupby("day").agg(volume=("q", "size"), empty=("empty", "mean"), p95=("ms", lambda s: s.quantile(0.95)),
                                 down=("feedback", lambda s: (s == "down").sum() / max(1, s.notna().sum())))


RULES = {"empty": 0.15, "p95": 2500, "down": 0.25}


def check(days, window=2):
    """Fire when a metric is over its threshold for `window` days in a row. Send these to chat or email."""
    out = []
    for metric, limit in RULES.items():
        over = (days[metric] > limit).astype(int)
        streak = over.groupby((over == 0).cumsum()).cumsum()
        for day in days.index[(streak == window)]:
            out.append((day, metric, float(days.loc[day, metric])))
    return sorted(out)


def drill(day, path=LOG):
    df = pd.read_json(path, lines=True)
    df["day"] = pd.to_datetime(df["ts"], unit="s").dt.date
    return df[(df["day"] == day) & df["empty"]]["q"].value_counts().head(5)


if __name__ == "__main__":
    days = daily()
    print(days.tail(7))
    for day, metric, value in check(days):
        print(f"ALERT {day} {metric} = {value:.2f}")
        if metric == "empty":
            print(drill(day))
```
