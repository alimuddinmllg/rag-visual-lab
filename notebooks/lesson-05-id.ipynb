```python
# NalarNusa RAG Visual Lab · Lesson 5 · Measuring similarity between two vectors
import math
import numpy as np

def from_angle(angle_deg, length):
    a = math.radians(angle_deg)
    return np.array([length * math.cos(a), length * math.sin(a)])

def dot(a, b):
    """Multiply matching numbers and add them up."""
    return float(np.sum(a * b))

def length(v):
    return float(np.sqrt(np.sum(v * v)))

def cosine(a, b):
    """Direction only: 1 = same direction, 0 = unrelated (90°), -1 = opposite."""
    return dot(a, b) / (length(a) * length(b))

def euclidean(a, b):
    """Straight-line distance between the two tips. Sensitive to length."""
    return length(a - b)

def run(angle_a, len_a, angle_b, len_b):
    a, b = from_angle(angle_a, len_a), from_angle(angle_b, len_b)
    res = {"dot": dot(a, b), "len_a": length(a), "len_b": length(b),
           "cosine": cosine(a, b), "distance": euclidean(a, b)}
    for k, v in res.items():
        print(f"{k:9s} {v:7.3f}")
    return res
```

```python
# NalarNusa RAG Visual Lab · Lesson 5 · Similarity with numpy, scikit-learn and sentence-transformers
# pip install numpy scikit-learn sentence-transformers
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity, euclidean_distances
from sentence_transformers import SentenceTransformer, util

model = SentenceTransformer("sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2")
a, b = model.encode(["How much does printing cost?", "PLA prints cost Rp 2.000 per gram."])

print("dot     ", float(np.dot(a, b)))

print("lengths ", float(np.linalg.norm(a)), float(np.linalg.norm(b)))

print("cosine  ", float(cosine_similarity([a], [b])[0, 0]), float(util.cos_sim(a, b)))

print("distance", float(euclidean_distances([a], [b])[0, 0]))
# After normalize_embeddings=True all three rank results the same way.
```
