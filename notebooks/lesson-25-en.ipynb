## Part 1 · From scratch

Code from lesson 11 (`ann.py`), which this lesson builds on.

```python
# NalarNusa RAG Visual Lab · Lesson 11 · Exact search vs a graph index (NSW, the idea behind HNSW)

def rng(seed):
    s = seed % 2147483646 + 1
    def rand():
        nonlocal s
        s = s * 16807 % 2147483647
        return s / 2147483647
    return rand

def make_points(n, seed, clusters=6):
    r = rng(seed)
    centers = [(0.12 + r() * 0.76, 0.12 + r() * 0.76) for _ in range(clusters)]
    pts = []
    for i in range(n):
        cx, cy = centers[i % clusters]
        jx = (r() + r() + r()) / 3 - 0.5
        jy = (r() + r() + r()) / 3 - 0.5
        pts.append((min(1, max(0, cx + jx * 0.4)), min(1, max(0, cy + jy * 0.4))))
    return pts

def d2(a, b):
    return (a[0] - b[0]) ** 2 + (a[1] - b[1]) ** 2

def build_nsw(points, M):
    """Insert points one by one; link each to its M nearest earlier points, both ways."""
    nbrs = [[] for _ in points]
    for i in range(1, len(points)):
        cand = sorted((d2(points[i], points[j]), j) for j in range(i))
        for _, j in cand[:M]:
            nbrs[i].append(j)
            nbrs[j].append(i)
    return nbrs

def brute_force(points, q, k):
    """Compare the query with every single vector: exact, but cost grows with the collection."""
    return [i for _, i in sorted((d2(p, q), i) for i, p in enumerate(points))[:k]], len(points)

def search_nsw(points, nbrs, q, k, ef, entry=0):
    """Walk the graph: always expand the closest unexplored point, stop when nothing closer is left."""
    count = 0
    def dist(i):
        nonlocal count
        count += 1
        return d2(points[i], q)
    visited = {entry}
    d0 = dist(entry)
    cand, res = [(d0, entry)], [(d0, entry)]
    width = max(ef, k)
    while cand:
        cand.sort()
        d, c = cand.pop(0)
        if len(res) >= width and d > res[-1][0]:
            break
        for nb in nbrs[c]:
            if nb in visited:
                continue
            visited.add(nb)
            dn = dist(nb)
            if len(res) < width or dn < res[-1][0]:
                cand.append((dn, nb))
                res.append((dn, nb))
                res.sort()
                if len(res) > width:
                    res.pop()
    return [i for _, i in res[:k]], count

def run(n, seed, M, ef, qx, qy, k=5):
    pts = make_points(n, seed)
    nbrs = build_nsw(pts, M)
    exact, c1 = brute_force(pts, (qx, qy), k)
    found, c2 = search_nsw(pts, nbrs, (qx, qy), k, ef)
    recall = len(set(exact) & set(found)) / k
    print(f"brute force: {c1} comparisons; graph: {c2} comparisons; recall@{k} = {recall:.0%}")
    return {"exact": exact, "found": found, "brute": c1, "graph": c2}
```

This lesson's code (`ivf.py`).

```python
# NalarNusa RAG Visual Lab · Lesson 25 · Scaling: an IVF index, its recall/latency curve, and memory at millions
# IVF = "inverted file": group the vectors into cells, then search only the few cells nearest the query.

def make_queries(points, seed, count=40):
    """Queries near stored vectors, like real questions near real chunks."""
    r = rng(seed + 99)
    out = []
    for j in range(count):
        x, y = points[(j * 37) % len(points)]
        out.append((x + (r() - 0.5) * 0.2, y + (r() - 0.5) * 0.2))
    return out

def brute_force(points, q, k):
    return [i for _, i in sorted((d2(p, q), i) for i, p in enumerate(points))[:k]]

def nearest(centroids, p):
    best, bd = 0, d2(centroids[0], p)
    for c in range(1, len(centroids)):
        d = d2(centroids[c], p)
        if d < bd:
            best, bd = c, d
    return best

def kmeans(points, nlist, iters=8):
    """Plain k-means: start from evenly spaced points, then move each centre to the mean of its points."""
    centroids = [points[(i * len(points)) // nlist] for i in range(nlist)]
    for _ in range(iters):
        sx, sy, cnt = [0.0] * nlist, [0.0] * nlist, [0] * nlist
        for p in points:
            c = nearest(centroids, p)
            sx[c] += p[0]; sy[c] += p[1]; cnt[c] += 1
        centroids = [(sx[c] / cnt[c], sy[c] / cnt[c]) if cnt[c] else centroids[c] for c in range(nlist)]
    return centroids

def inverted_lists(points, centroids):
    """Each cell keeps the ids of its vectors. This is the index."""
    lists = [[] for _ in centroids]
    for i, p in enumerate(points):
        lists[nearest(centroids, p)].append(i)
    return lists

def sweep(points, centroids, lists, queries, exact, k):
    """For nprobe = 1..nlist: recall@k and comparisons, averaged over the queries."""
    nlist = len(centroids)
    recall, comps = [0.0] * nlist, [0] * nlist
    for q, truth in zip(queries, exact):
        order = [c for _, c in sorted((d2(centroids[c], q), c) for c in range(nlist))]
        best, scanned = [], 0
        for p in range(nlist):                       # probe one more cell each time
            cell = lists[order[p]]
            scanned += len(cell)
            best = sorted(best + [(d2(points[i], q), i) for i in cell])[:k]
            recall[p] += len({i for _, i in best} & set(truth)) / k
            comps[p] += nlist + scanned              # centroid comparisons + vectors scanned
    n = len(queries)
    return [(p + 1, recall[p] / n, comps[p] / n) for p in range(nlist)]

def choose(curve, target, n):
    """The smallest nprobe that reaches the target recall."""
    for nprobe, rec, comp in curve:
        if rec >= target:
            return nprobe, rec, comp, n / comp
    nprobe, rec, comp = curve[-1]
    return nprobe, rec, comp, n / comp

def at_scale(millions, dims, precision, fraction):
    """Memory and time per query when the collection has millions of vectors (rough CPU: 1e10 multiply-adds/s)."""
    N = millions * 1_000_000
    gb = N * dims * (4 if precision == "float32" else 1) / 1e9
    exact_ms = N * dims / 1e10 * 1000
    return gb, exact_ms, exact_ms * fraction

def demo(n, nlist, target, millions, dims, precision, seed=7, k=10):
    points = make_points(n, seed)
    queries = make_queries(points, seed)
    exact = [brute_force(points, q, k) for q in queries]
    centroids = kmeans(points, nlist)
    lists = inverted_lists(points, centroids)
    curve = sweep(points, centroids, lists, queries, exact, k)
    nprobe, rec, comp, speedup = choose(curve, target, n)
    gb, exact_ms, ivf_ms = at_scale(millions, dims, precision, comp / n)
    for p, r, c in curve:
        print(f"nprobe {p:2d}: recall@{k} {r:.0%}  comparisons {c:7.1f}")
    print(f"pick nprobe = {nprobe}: recall {rec:.0%}, {speedup:.1f}x fewer comparisons than brute force")
    print(f"{millions}M x {dims} dims ({precision}): {gb:.1f} GB, exact {exact_ms:.0f} ms vs IVF {ivf_ms:.1f} ms per query")
    return {"sizes": [len(l) for l in lists], "curve": curve, "nprobe": nprobe, "gb": gb, "exact_ms": exact_ms, "ivf_ms": ivf_ms}
```

Run it with the settings from the Data cell:

```python
result = demo(N, NLIST, TARGET, MILLIONS, DIMS, PRECISION)
```

```python
# NalarNusa RAG Visual Lab · Lesson 25 · IVF with FAISS: the nprobe curve on embedding-sized data, and int8 storage
# pip install faiss-cpu numpy
import time
import faiss
import numpy as np

rng = np.random.default_rng(7)
D, N = 384, 200_000
centres = rng.normal(size=(500, D)).astype(np.float32)             # real embeddings form topic clusters
X = centres[rng.integers(0, 500, N)] + 0.45 * rng.normal(size=(N, D)).astype(np.float32)
faiss.normalize_L2(X)
Q = X[rng.integers(0, N, 50)] + 0.1 * rng.normal(size=(50, D)).astype(np.float32)
faiss.normalize_L2(Q)
flat = faiss.IndexFlatL2(D); flat.add(X)
t = time.perf_counter(); _, exact = flat.search(Q, 10); t_exact = (time.perf_counter() - t) / len(Q)

nlist = int(np.sqrt(N))                                            # rule of thumb: about sqrt(N) cells
quantizer = faiss.IndexFlatL2(D)                                  # finds the nearest cell centres
ivf = faiss.IndexIVFFlat(quantizer, D, nlist)
ivf.train(X[:50_000])                                              # k-means on a sample is enough

ivf.add(X)                                                         # every vector goes into its nearest cell's list

for nprobe in (1, 2, 4, 8, 16, 32, 64):
    ivf.nprobe = nprobe
    t = time.perf_counter(); _, found = ivf.search(Q, 10); ms = (time.perf_counter() - t) / len(Q) * 1e3
    recall = np.mean([len(set(a) & set(b)) / 10 for a, b in zip(found, exact)])
    print(f"nprobe {nprobe:2d}: recall@10 {recall:.0%}  {ms:.2f} ms/query  (exact {t_exact * 1e3:.2f} ms)")

# Same index, vectors stored as 8-bit integers: about 4x less memory, a little less accurate.
quantizer8 = faiss.IndexFlatL2(D)
sq = faiss.IndexIVFScalarQuantizer(quantizer8, D, nlist, faiss.ScalarQuantizer.QT_8bit)
sq.train(X[:50_000]); sq.add(X); sq.nprobe = 16
_, found = sq.search(Q, 10)
print(f"int8 recall@10 at nprobe 16: {np.mean([len(set(a) & set(b)) / 10 for a, b in zip(found, exact)]):.0%}")
print(f"memory for the vectors: float32 {N * D * 4 / 1e6:.0f} MB, int8 {N * D / 1e6:.0f} MB")
```
