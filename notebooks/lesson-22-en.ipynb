```python
# NalarNusa RAG Visual Lab · Lesson 6 · A RAG pipeline from scratch
# Only Python + numpy. Each stage below is the code the visual runs.
import math
import re
import numpy as np

def parse_document(text):
    """Split the document into words and remember which section each word is in."""
    sections, words = [], []
    for line in text.split("\n"):
        t = line.strip()
        if not t:
            continue
        if t.startswith("## "):
            sections.append(t[3:])
            continue
        for w in t.split():
            words.append((w, max(0, len(sections) - 1)))
    if not sections:
        sections.append("Document")
    return sections, words

def chunk_document(words, size, overlap):
    """Cut the word list into chunks of `size` words that overlap by `overlap` words."""
    step = max(5, size - overlap)
    chunks = []
    start = 0
    while start < len(words):
        ws = words[start:start + size]
        counts = {}
        for _, sec in ws:
            counts[sec] = counts.get(sec, 0) + 1
        ranked = sorted(counts.items(), key=lambda kv: (-kv[1], kv[0]))
        mixed = len(ranked) > 1 and ranked[0][1] / len(ws) < 0.75
        chunks.append({
            "id": len(chunks),
            "text": " ".join(w for w, _ in ws),
            "sec": ranked[0][0],
            "sec2": ranked[1][0] if mixed else None,
        })
        if start + size >= len(words):
            break
        start += step
    return chunks

def stem(w, lang):
    """A tiny English stemmer: printing -> print, prints -> print."""
    if lang != "en":
        return w
    if len(w) > 5 and w.endswith("ing"):
        return w[:-3]
    if len(w) > 4 and w.endswith("ies"):
        return w[:-3] + "y"
    if len(w) > 3 and w.endswith("s") and not w.endswith("ss"):
        return w[:-1]
    return w

def tokenize(text, stop, lang):
    words = re.findall(r"[a-z0-9]+", text.lower())
    return [stem(t, lang) for t in words if len(t) > 1 and t not in stop]

def vectorize(tokens, idf):
    """TF-IDF vector, scaled to length 1. Stored as {word: weight}."""
    tf = {}
    for t in tokens:
        if t in idf:
            tf[t] = tf.get(t, 0) + 1
    v = {t: (1 + math.log(c)) * idf[t] for t, c in tf.items()}
    norm = math.sqrt(sum(w * w for w in v.values())) or 1
    return {t: w / norm for t, w in v.items()}

def embed_chunks(chunks, stop, lang):
    toks = [tokenize(c["text"], stop, lang) for c in chunks]
    df = {}
    for ts in toks:
        for t in dict.fromkeys(ts):          # each word once per chunk
            df[t] = df.get(t, 0) + 1
    n = len(chunks)
    idf = {t: math.log((n + 1) / (d + 1)) + 1 for t, d in df.items()}
    vocab = list(idf.keys())
    vectors = [vectorize(ts, idf) for ts in toks]
    return idf, vocab, vectors

def pca_2d(vectors, vocab):
    """Squash the vectors to 2D for the map (PCA by power iteration)."""
    index = {t: i for i, t in enumerate(vocab)}
    X = np.zeros((len(vectors), len(vocab)))
    for i, v in enumerate(vectors):
        for t, w in v.items():
            X[i, index[t]] = w
    mean = X.mean(axis=0)
    Xc = X - mean
    K = Xc @ Xc.T
    comps = []
    for c in range(2):
        v = np.array([1 + ((i * 7 + c * 3) % 5) * 0.13 for i in range(len(vectors))])
        lam = 0.0
        for _ in range(150):
            w = K @ v
            nm = np.linalg.norm(w) or 1
            v, lam = w / nm, nm
        if v[np.argmax(np.abs(v))] < 0:
            v = -v
        comps.append((v, lam))
        K = K - lam * np.outer(v, v)
    coords = np.stack([v * math.sqrt(max(lam, 1e-12)) for v, lam in comps], axis=1)
    axes = [Xc.T @ (v / math.sqrt(max(lam, 1e-12))) for v, lam in comps]
    return coords, mean, axes, index

def embed_question(question, idf, stop, lang):
    q_tokens = tokenize(question, stop, lang)
    qv = vectorize(q_tokens, idf)
    known = list(dict.fromkeys(t for t in q_tokens if t in idf))
    return qv, known

def cosine(a, b):
    """Both vectors have length 1, so cosine similarity is just the dot product."""
    if len(a) > len(b):
        a, b = b, a
    return sum(w * b[t] for t, w in a.items() if t in b)

def retrieve(qv, vectors, top_k):
    scores = [cosine(qv, v) if qv else 0.0 for v in vectors]
    order = sorted(range(len(scores)), key=lambda i: -scores[i])
    return scores, [(i, scores[i]) for i in order[:top_k]]

PROMPT = {
    "en": dict(
        system="You are a helpful assistant for the NusaBot Robotics Lab.\n"
               "Answer ONLY using the context below and cite sources like [1].\n"
               "If the answer is not in the context, say you don't know.",
        context="Context:", none="(no relevant chunks found)", question="Question:", answer="Answer:",
        refuse="I don't know. The document doesn't contain anything about this question."),
    "id": dict(
        system="Anda adalah asisten untuk NusaBot Robotics Lab.\n"
               "Jawab HANYA berdasarkan konteks di bawah dan sebutkan sumber seperti [1].\n"
               "Jika jawabannya tidak ada di konteks, katakan Anda tidak tahu.",
        context="Konteks:", none="(tidak ada potongan yang relevan)", question="Pertanyaan:", answer="Jawaban:",
        refuse="Saya tidak tahu. Dokumen tidak berisi informasi tentang pertanyaan ini."),
}

def build_prompt(question, used, chunks, sections, lang):
    T = PROMPT.get(lang, PROMPT["en"])
    ctx = "\n\n".join(f"[{r + 1}] ({sections[chunks[i]['sec']]}) {chunks[i]['text']}"
                      for r, (i, _) in enumerate(used))
    return f"{T['system']}\n\n{T['context']}\n{ctx or T['none']}\n\n{T['question']} {question}\n{T['answer']}"

def extract_answer(used, chunks, known, idf, stop, lang):
    """Stand-in for an LLM: pick the best-matching sentences and cite them."""
    T = PROMPT.get(lang, PROMPT["en"])
    if not used:
        return T["refuse"]
    qset = set(known)
    cands = []
    for r, (i, _) in enumerate(used):
        for sent in re.split(r"(?<=[.!?])\s+", chunks[i]["text"]):
            sc = sum(idf.get(x, 0) for x in dict.fromkeys(tokenize(sent, stop, lang)) if x in qset)
            if sc > 0:
                cands.append((sent, sc, r))
    cands.sort(key=lambda c: -c[1])
    picks = []
    for sent, sc, r in cands:
        if len(picks) >= 2:
            break
        if len(re.split(r"\s+", sent)) < 5:
            continue
        if any(sent[:40] in p[0] or p[0][:40] in sent for p in picks):   # overlap repeats text
            continue
        if picks and sc < picks[0][1] * 0.6:
            break
        picks.append((sent, sc, r))
    if not picks:
        return T["refuse"]
    fmt = lambda s: ("" if re.match(r"[A-Z0-9]", s) else "…") + s + ("" if re.search(r"[.!?]$", s) else "…")
    return " ".join(f"{fmt(s)} [{r + 1}]" for s, _, r in picks)

def run_pipeline(document, question, size, overlap, top_k, min_sim, lang, stop, verbose=False):
    log = print if verbose else (lambda *a: None)
    sections, words = parse_document(document)
    chunks = chunk_document(words, size, overlap)
    log(f"chunk    size={size} overlap={overlap} -> {len(chunks)} chunks")

    idf, vocab, vectors = embed_chunks(chunks, stop, lang)
    coords, mean, axes, _ = pca_2d(vectors, vocab)
    log(f"embed    TF-IDF, {len(vocab)} terms -> {len(vectors)} vectors")

    qv, known = embed_question(question, idf, stop, lang)
    log(f"query    known words: {', '.join(known) or '(none)'}")

    scores, top = retrieve(qv, vectors, top_k)
    used = [(i, s) for i, s in top if s >= min_sim and known]
    log("retrieve " + ", ".join(f"#{i + 1} {s:.2f}" for i, s in top))

    prompt = build_prompt(question, used, chunks, sections, lang)
    answer = extract_answer(used, chunks, known, idf, stop, lang)
    log(f"generate {len(used)} chunks in prompt\n\nAnswer: {answer}")

    query_xy = None
    if known:
        q = np.zeros(len(vocab))
        for t, w in qv.items():
            q[vocab.index(t)] = w
        query_xy = [float((q - mean) @ a) for a in axes]
    return {
        "n_chunks": len(chunks),
        "vocab_size": len(vocab),
        "known": known,
        "top": [[i, s] for i, s in top],
        "used": [i for i, _ in used],
        "coords": coords.tolist(),
        "query_xy": query_xy,
        "prompt": prompt,
        "answer": answer,
    }
```

Code from lesson 13 (`bm25.py`), which this lesson builds on.

```python
# NalarNusa RAG Visual Lab · Lesson 13 · BM25 keyword search vs vector search
import math

def index(document, lang, stop, size=40, overlap=8):
    _, words = parse_document(document)
    chunks = chunk_document(words, size, min(overlap, size - 6))
    toks = [tokenize(c["text"], stop, lang) for c in chunks]
    return chunks, toks

def bm25_scores(toks, q_tokens, k1=1.2, b=0.75):
    """Rewards rare query words that appear often in a chunk, with diminishing returns (k1)
    and a penalty for long chunks (b)."""
    N = len(toks)
    df = {}
    for ts in toks:
        for t in set(ts):
            df[t] = df.get(t, 0) + 1
    avgdl = sum(len(t) for t in toks) / max(1, N)
    scores = []
    for ts in toks:
        tf = {}
        for t in ts:
            tf[t] = tf.get(t, 0) + 1
        s = 0.0
        for t in dict.fromkeys(q_tokens):
            f = tf.get(t)
            if not f:
                continue
            idf = math.log((N - df[t] + 0.5) / (df[t] + 0.5) + 1)
            s += idf * (f * (k1 + 1)) / (f + k1 * (1 - b + b * len(ts) / avgdl))
        scores.append(s)
    return scores

def vector_scores(chunks, question, stop, lang):
    idf, _, vectors = embed_chunks(chunks, stop, lang)
    qv, _ = embed_question(question, idf, stop, lang)
    return [cosine(qv, v) if qv else 0.0 for v in vectors]

def rank(scores):
    return sorted(range(len(scores)), key=lambda i: (-scores[i], i))

def demo(document, question, lang, stop, k1=1.2, b=0.75):
    stop = set(stop)
    chunks, toks = index(document, lang, stop)
    q_tokens = tokenize(question, stop, lang)
    bm = bm25_scores(toks, q_tokens, k1, b)
    vec = vector_scores(chunks, question, stop, lang)
    rb, rv = rank(bm), rank(vec)
    print("query words:", q_tokens)
    print("BM25  :", ", ".join(f"#{i + 1} {bm[i]:.2f}" for i in rb[:5]))
    print("vector:", ", ".join(f"#{i + 1} {vec[i]:.2f}" for i in rv[:5]))
    shared = len(set(rb[:5]) & set(rv[:5]))
    print(f"shared in top 5: {shared}/5   same #1: {rb[0] == rv[0]}")
    return {"bm25": rb[:5], "vector": rv[:5]}
```

Code from lesson 14 (`fusion.py`), which this lesson builds on.

```python
# NalarNusa RAG Visual Lab · Lesson 14 · Hybrid search with reciprocal rank fusion (RRF)
import re

def keyword_ranking(chunks, toks, question, stop, lang):
    return rank(bm25_scores(toks, tokenize(question, stop, lang)))

def vector_ranking(chunks, question, stop, lang):
    return rank(vector_scores(chunks, question, stop, lang))

def rrf(rankings, k=60, depth=20):
    """Each list votes 1 / (k + rank) for its items. Agreement between lists wins;
    raw scores are ignored, so BM25 and cosine never need to be on the same scale."""
    score = {}
    for r in rankings:
        for pos, i in enumerate(r[:depth]):
            score[i] = score.get(i, 0.0) + 1 / (k + pos + 1)
    return sorted(score, key=lambda i: (-score[i], i))

def hit_rate(rankings_per_question, relevant_per_question, k):
    hits = sum(any(i in rel for i in r[:k]) for r, rel in zip(rankings_per_question, relevant_per_question))
    return hits / len(relevant_per_question)

def norm(s):
    return " " + re.sub(r"[^a-z0-9]+", " ", s.lower()).strip() + " "

def demo(document, question, tests, lang, stop, rrf_k=60, k=1):
    stop = set(stop)
    chunks, toks = index(document, lang, stop)
    kw, vec = keyword_ranking(chunks, toks, question, stop, lang), vector_ranking(chunks, question, stop, lang)
    fused = rrf([kw, vec], rrf_k)
    print("keyword:", [i + 1 for i in kw[:5]], " vector:", [i + 1 for i in vec[:5]], " fused:", [i + 1 for i in fused[:5]])
    if not tests:
        print("No test set for this document: build one in lesson 19.")
        return {"fused": fused[:5], "rates": None}
    rel = [{i for i, c in enumerate(chunks) if norm(e) in norm(c["text"])} for _, e in tests]
    runs = {"bm25": [], "vector": [], "hybrid": []}
    for q, _ in tests:
        b, v = keyword_ranking(chunks, toks, q, stop, lang), vector_ranking(chunks, q, stop, lang)
        runs["bm25"].append(b); runs["vector"].append(v); runs["hybrid"].append(rrf([b, v], rrf_k))
    rates = {name: hit_rate(rs, rel, k) for name, rs in runs.items()}
    print(f"hit@{k} over {len(tests)} questions:", "  ".join(f"{n} {r:.0%}" for n, r in rates.items()))
    return {"fused": fused[:5], "rates": rates}
```

```python
# NalarNusa RAG Visual Lab · Lesson 16 · Query rewriting, multi-query and HyDE
# A synonym table stands in for an LLM that rewrites questions; the tiny model from
# lesson 1 writes the HyDE fake answer. The LLM version is in the "With libraries" tab.
import re

SYNONYMS = {
    "en": {
        "price": ["cost","costs"], "fee": ["cost","charged"], "fine": ["charged","late"],
        "penalty": ["charged","late"], "hours": ["open","closed"], "open": ["opening"],
        "borrow": ["loan","borrow"], "loan": ["borrow"], "rules": ["must","required"],
        "close": ["closed","closes"], "shut": ["closed","closes"], "weekend": ["saturday","sunday"],
        "computer": ["workstation","workstations"], "pc": ["workstation"], "phone": ["whatsapp"],
        "contact": ["front desk","whatsapp"], "battery": ["lithium","batteries"], "charge": ["charged","charging"],
        "member": ["membership"], "join": ["membership"], "print": ["printing","prints"],
    },
    "id": {
        "harga": ["biaya"], "tarif": ["biaya"], "ongkos": ["biaya"],
        "denda": ["dikenakan","terlambat"], "pinjam": ["meminjam"], "tutup": ["tutup","jam buka"],
        "buka": ["jam buka"], "aturan": ["wajib"], "akhir": ["sabtu","minggu"],
        "pekan": ["sabtu","minggu"], "komputer": ["komputer kerja"], "kontak": ["meja depan","whatsapp"],
        "hubungi": ["whatsapp","meja depan"], "baterai": ["litium"], "anggota": ["keanggotaan"],
        "daftar": ["keanggotaan"], "cetak": ["cetak","printer"],
    },
}

def search(chunks, query, stop, lang):
    return rank(vector_scores(chunks, query, stop, lang))       # TF-IDF ranking, best first

# ranking = search(chunks, question, stop, lang)

def rewrite_query(question, lang):
    """Versions of the question in the document's vocabulary."""
    table = SYNONYMS.get(lang, SYNONYMS["en"])
    words = re.findall(r"[a-z0-9]+", question.lower())
    variants = [question]
    swap = [w for w in words if w in table]
    if swap:
        variants.append(" ".join(table[w][0] if w in table else w for w in words))
        extra = list(dict.fromkeys(x for w in swap for x in table[w]))
        variants.append(question + " " + " ".join(extra))
    return variants

def multi_query(chunks, question, stop, lang):
    return rrf([search(chunks, v, stop, lang) for v in rewrite_query(question, lang)])

def hyde(question, model):
    """Write a fake answer, search with it. Its facts may be wrong; it only has to SOUND like the documents."""
    seed = [t for t in lm_tokens(question) if re.search(r"[a-z]", t)][-2:]
    return question + " " + detok(seed + generate(model, seed, 14, 0.7, 5))

def norm(s):
    return " " + re.sub(r"[^a-z0-9]+", " ", s.lower()).strip() + " "

def demo(document, general, question, paraphrases, lang, stop, k=1):
    stop = set(stop)
    chunks, _ = index(document, lang, stop)
    model = train_lm(lm_tokens(general))
    fake = hyde(question, model)
    print("versions:", rewrite_query(question, lang))
    print("fake answer:", fake)
    orig, multi, hy = search(chunks, question, stop, lang), multi_query(chunks, question, stop, lang), search(chunks, fake, stop, lang)
    print("top 3  original", [i + 1 for i in orig[:3]], " multi-query", [i + 1 for i in multi[:3]], " HyDE", [i + 1 for i in hy[:3]])
    if not paraphrases:
        return {"orig": orig[:3], "multi": multi[:3], "hyde": hy[:3], "rates": None}
    s = {"original": 0, "multi": 0, "hyde": 0}
    for q, e in paraphrases:
        rel = {i for i, c in enumerate(chunks) if norm(e) in norm(c["text"])}
        s["original"] += any(i in rel for i in search(chunks, q, stop, lang)[:k])
        s["multi"] += any(i in rel for i in multi_query(chunks, q, stop, lang)[:k])
        s["hyde"] += any(i in rel for i in search(chunks, hyde(q, model), stop, lang)[:k])
    rates = {n: v / len(paraphrases) for n, v in s.items()}
    print(f"hit@{k} on {len(paraphrases)} paraphrased questions:", "  ".join(f"{n} {r:.0%}" for n, r in rates.items()))
    return {"orig": orig[:3], "multi": multi[:3], "hyde": hy[:3], "rates": rates}
```

This lesson's code (`debug.py`).

```python
# NalarNusa RAG Visual Lab · Lesson 22 · Failure clinic: debug RAG stage by stage, not by guessing
import re

# Six broken pipelines. Each has one faulty setting and its fix.
CASES = [
    {"id": 1, "cause": "chunking", "q": {"en": "How much does PLA printing cost?", "id": "Berapa biaya cetak PLA?"},
     "expect": {"en": "2.000 per gram", "id": "2.000 per gram"}, "broken": {"size": 6, "overlap": 0, "top_k": 1}, "fixed": {"size": 40, "overlap": 8, "top_k": 3}},
    {"id": 2, "cause": "threshold", "q": {"en": "When is the lab open on Saturday?", "id": "Kapan lab buka hari Sabtu?"},
     "expect": {"en": "Saturday from 10:00", "id": "Sabtu pukul 10.00"}, "broken": {"min_sim": 0.6}, "fixed": {"min_sim": 0.1}},
    {"id": 3, "cause": "filter", "q": {"en": "Where must lithium batteries be charged?", "id": "Di mana baterai litium harus diisi?"},
     "expect": {"en": "fireproof charging bags", "id": "kantong pengisian tahan api"}, "broken": {"filter": 2}, "fixed": {"filter": None}},
    {"id": 4, "cause": "topk", "q": {"en": "How much does printing cost?", "id": "Berapa biaya mencetak?"}, "in_chunks": True,
     "expect": {"en": "500 per page", "id": "2.000 per gram"}, "broken": {"top_k": 1}, "fixed": {"top_k": 3}},
    {"id": 5, "cause": "language", "q": {"en": "When does the lab close during exam weeks?", "id": "When does the lab close during exam weeks?"},
     "expect": {"en": "18:00", "id": "18.00"}, "broken": {"doc_lang": "other"}, "fixed": {"doc_lang": "same"}},
    {"id": 6, "cause": "wording", "q": {"en": "What is the fee to join as a student?", "id": "Berapa ongkos daftar anggota mahasiswa?"},
     "expect": {"en": "50.000 per semester", "id": "50.000 per semester"}, "broken": {"rewrite": False}, "fixed": {"rewrite": True}},
]

def norm(s):
    return " " + re.sub(r"[^a-z0-9]+", " ", s.lower()).strip() + " "

def run_case(case, lang, which, handbook, stopwords):
    """Run the whole pipeline with this case's settings and log every stage."""
    p = {"size": 40, "overlap": 8, "top_k": 3, "min_sim": 0.1, "filter": None, "doc_lang": "same", "rewrite": False, **case[which]}
    doc_lang = ("en" if p["doc_lang"] == "same" else "id") if case["id"] == 5 else lang
    question = case["q"]["en"] if case["id"] == 5 else case["q"][lang]
    stop = set(stopwords[doc_lang])
    chunks, _ = index(handbook[doc_lang], doc_lang, stop, p["size"], p["overlap"])
    idf, _, _ = embed_chunks(chunks, stop, doc_lang)
    variants = rewrite_query(question, doc_lang) if p["rewrite"] else [question]
    cos = vector_scores(chunks, question, stop, doc_lang)
    if len(variants) > 1:
        ranking = rrf([rank(vector_scores(chunks, v, stop, doc_lang)) for v in variants])
    else:
        ranking = rank(cos)
    if p["filter"] is not None:
        ranking = [i for i in ranking if chunks[i]["sec"] == p["filter"]]
    top = [(i, max(cos[i], 0.11) if len(variants) > 1 else cos[i]) for i in ranking[:p["top_k"]]]
    used = [(i, s) for i, s in top if s >= p["min_sim"]]
    known = [t for v in variants for t in tokenize(v, stop, doc_lang) if t in idf]
    answer = extract_answer(used, chunks, known, idf, stop, doc_lang)
    expect = ("18:00" if doc_lang == "en" else "18.00") if case["id"] == 5 else case["expect"][lang]
    in_prompt = any(norm(expect) in norm(chunks[i]["text"]) for i, _ in used)
    ok = in_prompt if case.get("in_chunks") else norm(expect) in norm(answer)
    return {"question": question, "settings": p, "top": top, "used": [i for i, _ in used],
            "answer_in_prompt": in_prompt, "answer": answer, "ok": ok}

def diagnose(r):
    """Work backwards from the answer."""
    if r["answer_in_prompt"] and not r["ok"]:
        return "generation: the answer was in the prompt; check the prompt or the model (lesson 21)"
    if not r["used"] and r["top"]:
        return "retrieved but dropped: similarity threshold, top-k or token budget"
    return "not retrieved: chunking, a metadata filter, the question's wording, or the document's language"

# Change ONE thing, run again, and keep it only if the test set (lesson 20) agrees.

def demo(case_index, lang, handbook, stopwords):
    case = CASES[case_index]
    for which in ("broken", "fixed"):
        r = run_case(case, lang, which, handbook, stopwords)
        print(f"{which:6s} {'✓' if r['ok'] else '✗'}  top {[(i + 1, round(s, 2)) for i, s in r['top']]}  used {len(r['used'])}  -> {r['answer'][:70]}")
        if which == "broken":
            print("        diagnosis:", diagnose(r))
    print("\nall cases:", " ".join(f"{c['id']}:{'✓' if run_case(c, lang, 'fixed', handbook, stopwords)['ok'] else '✗'}" for c in CASES))
    return {w: run_case(case, lang, w, handbook, stopwords)["ok"] for w in ("broken", "fixed")} | {
        "answers": [run_case(case, lang, w, handbook, stopwords)["answer"] for w in ("broken", "fixed")]}
```

```python
# NalarNusa RAG Visual Lab · Lesson 22 · Log every stage with Langfuse-style tracing (here: plain logging)
# Production tools (Langfuse, Phoenix, LangSmith) record the same things; the habit matters more than the tool.
import json
import logging
import time

log = logging.getLogger("rag_trace")
log.setLevel(logging.INFO)
if not log.handlers:
    h = logging.FileHandler("rag_trace.jsonl", encoding="utf-8"); h.setFormatter(logging.Formatter("%(message)s")); log.addHandler(h)
log.propagate = False


def trace(stage, **data):
    log.info(json.dumps({"t": time.time(), "stage": stage, **data}, ensure_ascii=False))
    for handler in log.handlers:
        handler.flush()


def rag(question, retriever, llm, top_k=3, min_sim=0.1):
    hits = retriever(question, k=top_k)                               # [(chunk_id, score, text)]
    trace("retrieve", question=question, hits=[(i, round(s, 3)) for i, s, _ in hits])
    used = [h for h in hits if h[1] >= min_sim]
    trace("filter", kept=[h[0] for h in used], dropped=[h[0] for h in hits if h not in used])
    prompt = "\n\n".join(f"[{n}] {t}" for n, (_, _, t) in enumerate(used, 1)) + f"\n\nQuestion: {question}"
    answer = llm(prompt)
    trace("generate", prompt_chars=len(prompt), answer=answer)
    return answer

# Read rag_trace.jsonl for a wrong answer: was the expected text in "hits"? in "kept"? in the prompt?


# Try it with stand-ins, then plug in your real retriever and LLM.
fake_retriever = lambda q, k: [(7, 0.52, "Saturday from 10:00 to 16:00."), (3, 0.08, "Wear safety glasses.")][:k]
fake_llm = lambda prompt: "The lab is open on Saturday from 10:00 to 16:00 [1]."
print(rag("When is the lab open on Saturday?", fake_retriever, fake_llm))
print(open("rag_trace.jsonl", encoding="utf-8").read())
# Re-run the whole test set after each fix and compare the scores (lesson 20).
```
