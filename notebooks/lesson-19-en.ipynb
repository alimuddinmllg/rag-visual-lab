## Part 1 · From scratch

This lesson's code (`test_set.py`).

```python
# NalarNusa RAG Visual Lab · Lesson 19 · Draft a test set from your own document
# A test set is just data: questions plus a phrase the right answer must contain.
import json
import re

FACT = re.compile(
    r"(Rp\s?[0-9][0-9.,]*(?:\s+per\s+\w+)?"                       # prices
    r"|\b[0-9]{1,2}[:.][0-9]{2}\b"                                 # times
    r"|\b(?:one|two|three|four|five|six|seven|eight|nine|ten|satu|dua|tiga|empat|lima|enam|tujuh|delapan|sembilan|sepuluh)\s+\w+"
    r"|\b[0-9]+(?:[.,][0-9]+)?\s*(?:%|days?|hours?|weeks?|pages?|credits?|hari|jam|minggu|halaman|sks)\b)",
    re.I | re.A)

def find_facts(text, limit=15):
    out = []
    for line in text.split("\n"):
        t = line.strip()
        if not t or t.startswith("## "):
            continue
        for s in re.split(r"(?<=[.!?])\s+", t):
            m = FACT.search(s)
            if m and len(s.split()) >= 5:
                out.append((s, m.group(0)))
    return out[:limit]

def draft(facts, lang):
    head = "Menurut dokumen: " if lang == "id" else "According to the document: "
    return [[head + s.replace(f, "___", 1), f] for s, f in facts]

# Rewrite each draft the way a real person asks: "How much is a PLA print?" not
# "According to the document: PLA prints cost ___". Add questions the document can't answer.
# (The "With libraries" tab asks an LLM for a first rewrite; a person still checks it.)

def demo(text, lang, limit=15):
    drafts = draft(find_facts(text, limit), lang)
    print(json.dumps(drafts, ensure_ascii=False, indent=1))      # on your computer: json.dump(..., open("test_set.json", "w"))
    return {"drafts": drafts}
```

## Part 2 · With libraries

Installs the packages, then runs the library version. Cells that call an LLM expect an OpenAI-compatible server at `http://localhost:11434/v1` (Ollama on your own computer). In Colab, change `base_url` and `api_key` to a hosted service you have access to.

```python
# NalarNusa RAG Visual Lab · Lesson 19 · Let an LLM reword the drafts, then review them yourself
# pip install openai
import json
from openai import OpenAI

client = OpenAI(base_url="http://localhost:11434/v1", api_key="ollama")   # any OpenAI-compatible server

drafts = json.load(open("test_set.json", encoding="utf-8"))   # from test_set.py: [[draft question, fact], ...]

def reword(draft, fact):
    sentence = draft.split(": ", 1)[-1].replace("___", fact)
    r = client.chat.completions.create(model="qwen2.5:7b-instruct", temperature=0.7, messages=[{"role": "user", "content":
        "Write one short, natural question that a student would ask and that this sentence answers. "
        "Use different words from the sentence where you can. Reply with the question only, in the same "
        f"language as the sentence.\nSentence: {sentence}"}])
    return r.choices[0].message.content.strip()

reviewed = []
for q, fact in drafts:
    new = reword(q, fact)
    ok = input(f"{new}\n  expected: {fact}\n  keep? [y/n/edit] ")   # a person decides, every time
    if ok == "y":
        reviewed.append([new, fact])
    elif ok not in ("n", ""):
        reviewed.append([ok, fact])
reviewed += [["How do I renew my passport?", None]]               # unanswerable: the system should refuse

json.dump(reviewed, open("test_set.json", "w", encoding="utf-8"), ensure_ascii=False, indent=2)
```
