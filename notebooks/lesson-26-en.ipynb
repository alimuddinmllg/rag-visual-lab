```python
# NalarNusa RAG Visual Lab · Lesson 6 · A RAG pipeline from scratch
# Only Python + numpy. Each stage below is the code the visual runs.
import math
import re
import numpy as np

def parse_document(text):
    """Split the document into words and remember which section each word is in."""
    sections, words = [], []
    for line in text.split("\n"):
        t = line.strip()
        if not t:
            continue
        if t.startswith("## "):
            sections.append(t[3:])
            continue
        for w in t.split():
            words.append((w, max(0, len(sections) - 1)))
    if not sections:
        sections.append("Document")
    return sections, words

def chunk_document(words, size, overlap):
    """Cut the word list into chunks of `size` words that overlap by `overlap` words."""
    step = max(5, size - overlap)
    chunks = []
    start = 0
    while start < len(words):
        ws = words[start:start + size]
        counts = {}
        for _, sec in ws:
            counts[sec] = counts.get(sec, 0) + 1
        ranked = sorted(counts.items(), key=lambda kv: (-kv[1], kv[0]))
        mixed = len(ranked) > 1 and ranked[0][1] / len(ws) < 0.75
        chunks.append({
            "id": len(chunks),
            "text": " ".join(w for w, _ in ws),
            "sec": ranked[0][0],
            "sec2": ranked[1][0] if mixed else None,
        })
        if start + size >= len(words):
            break
        start += step
    return chunks

def stem(w, lang):
    """A tiny English stemmer: printing -> print, prints -> print."""
    if lang != "en":
        return w
    if len(w) > 5 and w.endswith("ing"):
        return w[:-3]
    if len(w) > 4 and w.endswith("ies"):
        return w[:-3] + "y"
    if len(w) > 3 and w.endswith("s") and not w.endswith("ss"):
        return w[:-1]
    return w

def tokenize(text, stop, lang):
    words = re.findall(r"[a-z0-9]+", text.lower())
    return [stem(t, lang) for t in words if len(t) > 1 and t not in stop]

def vectorize(tokens, idf):
    """TF-IDF vector, scaled to length 1. Stored as {word: weight}."""
    tf = {}
    for t in tokens:
        if t in idf:
            tf[t] = tf.get(t, 0) + 1
    v = {t: (1 + math.log(c)) * idf[t] for t, c in tf.items()}
    norm = math.sqrt(sum(w * w for w in v.values())) or 1
    return {t: w / norm for t, w in v.items()}

def embed_chunks(chunks, stop, lang):
    toks = [tokenize(c["text"], stop, lang) for c in chunks]
    df = {}
    for ts in toks:
        for t in dict.fromkeys(ts):          # each word once per chunk
            df[t] = df.get(t, 0) + 1
    n = len(chunks)
    idf = {t: math.log((n + 1) / (d + 1)) + 1 for t, d in df.items()}
    vocab = list(idf.keys())
    vectors = [vectorize(ts, idf) for ts in toks]
    return idf, vocab, vectors

def pca_2d(vectors, vocab):
    """Squash the vectors to 2D for the map (PCA by power iteration)."""
    index = {t: i for i, t in enumerate(vocab)}
    X = np.zeros((len(vectors), len(vocab)))
    for i, v in enumerate(vectors):
        for t, w in v.items():
            X[i, index[t]] = w
    mean = X.mean(axis=0)
    Xc = X - mean
    K = Xc @ Xc.T
    comps = []
    for c in range(2):
        v = np.array([1 + ((i * 7 + c * 3) % 5) * 0.13 for i in range(len(vectors))])
        lam = 0.0
        for _ in range(150):
            w = K @ v
            nm = np.linalg.norm(w) or 1
            v, lam = w / nm, nm
        if v[np.argmax(np.abs(v))] < 0:
            v = -v
        comps.append((v, lam))
        K = K - lam * np.outer(v, v)
    coords = np.stack([v * math.sqrt(max(lam, 1e-12)) for v, lam in comps], axis=1)
    axes = [Xc.T @ (v / math.sqrt(max(lam, 1e-12))) for v, lam in comps]
    return coords, mean, axes, index

def embed_question(question, idf, stop, lang):
    q_tokens = tokenize(question, stop, lang)
    qv = vectorize(q_tokens, idf)
    known = list(dict.fromkeys(t for t in q_tokens if t in idf))
    return qv, known

def cosine(a, b):
    """Both vectors have length 1, so cosine similarity is just the dot product."""
    if len(a) > len(b):
        a, b = b, a
    return sum(w * b[t] for t, w in a.items() if t in b)

def retrieve(qv, vectors, top_k):
    scores = [cosine(qv, v) if qv else 0.0 for v in vectors]
    order = sorted(range(len(scores)), key=lambda i: -scores[i])
    return scores, [(i, scores[i]) for i in order[:top_k]]

PROMPT = {
    "en": dict(
        system="You are a helpful assistant for the NusaBot Robotics Lab.\n"
               "Answer ONLY using the context below and cite sources like [1].\n"
               "If the answer is not in the context, say you don't know.",
        context="Context:", none="(no relevant chunks found)", question="Question:", answer="Answer:",
        refuse="I don't know. The document doesn't contain anything about this question."),
    "id": dict(
        system="Anda adalah asisten untuk NusaBot Robotics Lab.\n"
               "Jawab HANYA berdasarkan konteks di bawah dan sebutkan sumber seperti [1].\n"
               "Jika jawabannya tidak ada di konteks, katakan Anda tidak tahu.",
        context="Konteks:", none="(tidak ada potongan yang relevan)", question="Pertanyaan:", answer="Jawaban:",
        refuse="Saya tidak tahu. Dokumen tidak berisi informasi tentang pertanyaan ini."),
}

def build_prompt(question, used, chunks, sections, lang):
    T = PROMPT.get(lang, PROMPT["en"])
    ctx = "\n\n".join(f"[{r + 1}] ({sections[chunks[i]['sec']]}) {chunks[i]['text']}"
                      for r, (i, _) in enumerate(used))
    return f"{T['system']}\n\n{T['context']}\n{ctx or T['none']}\n\n{T['question']} {question}\n{T['answer']}"

def extract_answer(used, chunks, known, idf, stop, lang):
    """Stand-in for an LLM: pick the best-matching sentences and cite them."""
    T = PROMPT.get(lang, PROMPT["en"])
    if not used:
        return T["refuse"]
    qset = set(known)
    cands = []
    for r, (i, _) in enumerate(used):
        for sent in re.split(r"(?<=[.!?])\s+", chunks[i]["text"]):
            sc = sum(idf.get(x, 0) for x in dict.fromkeys(tokenize(sent, stop, lang)) if x in qset)
            if sc > 0:
                cands.append((sent, sc, r))
    cands.sort(key=lambda c: -c[1])
    picks = []
    for sent, sc, r in cands:
        if len(picks) >= 2:
            break
        if len(re.split(r"\s+", sent)) < 5:
            continue
        if any(sent[:40] in p[0] or p[0][:40] in sent for p in picks):   # overlap repeats text
            continue
        if picks and sc < picks[0][1] * 0.6:
            break
        picks.append((sent, sc, r))
    if not picks:
        return T["refuse"]
    fmt = lambda s: ("" if re.match(r"[A-Z0-9]", s) else "…") + s + ("" if re.search(r"[.!?]$", s) else "…")
    return " ".join(f"{fmt(s)} [{r + 1}]" for s, _, r in picks)

def run_pipeline(document, question, size, overlap, top_k, min_sim, lang, stop, verbose=False):
    log = print if verbose else (lambda *a: None)
    sections, words = parse_document(document)
    chunks = chunk_document(words, size, overlap)
    log(f"chunk    size={size} overlap={overlap} -> {len(chunks)} chunks")

    idf, vocab, vectors = embed_chunks(chunks, stop, lang)
    coords, mean, axes, _ = pca_2d(vectors, vocab)
    log(f"embed    TF-IDF, {len(vocab)} terms -> {len(vectors)} vectors")

    qv, known = embed_question(question, idf, stop, lang)
    log(f"query    known words: {', '.join(known) or '(none)'}")

    scores, top = retrieve(qv, vectors, top_k)
    used = [(i, s) for i, s in top if s >= min_sim and known]
    log("retrieve " + ", ".join(f"#{i + 1} {s:.2f}" for i, s in top))

    prompt = build_prompt(question, used, chunks, sections, lang)
    answer = extract_answer(used, chunks, known, idf, stop, lang)
    log(f"generate {len(used)} chunks in prompt\n\nAnswer: {answer}")

    query_xy = None
    if known:
        q = np.zeros(len(vocab))
        for t, w in qv.items():
            q[vocab.index(t)] = w
        query_xy = [float((q - mean) @ a) for a in axes]
    return {
        "n_chunks": len(chunks),
        "vocab_size": len(vocab),
        "known": known,
        "top": [[i, s] for i, s in top],
        "used": [i for i, _ in used],
        "coords": coords.tolist(),
        "query_xy": query_xy,
        "prompt": prompt,
        "answer": answer,
    }
```

```python
# NalarNusa RAG Visual Lab · Lesson 8 · Four ways to chunk a document
# Sizes are counted in words. Uses tokenize/vectorize/embed_chunks/cosine from lesson 6.
import re

def sentences_of(text):
    out, sec = [], -1
    for line in text.split("\n"):
        t = line.strip()
        if not t:
            continue
        if t.startswith("## "):
            sec += 1
            continue
        for s in re.split(r"(?<=[.!?])\s+", t):
            if s:
                out.append({"text": s, "words": len(s.split()), "sec": max(0, sec)})
    return out

def pack(sents, size):
    """Greedily fill chunks with whole sentences up to `size` words."""
    out, cur, n = [], [], 0
    def flush():
        nonlocal cur, n
        if cur:
            out.append({"text": " ".join(s["text"] for s in cur), "words": n, "sec": cur[0]["sec"]})
            cur, n = [], 0
    for s in sents:
        if s["words"] > size:                       # one long sentence: cut into fixed pieces
            flush()
            w = s["text"].split()
            for i in range(0, len(w), size):
                out.append({"text": " ".join(w[i:i + size]), "words": min(size, len(w) - i), "sec": s["sec"]})
            continue
        if n + s["words"] > size:
            flush()
        cur.append(s); n += s["words"]
    flush()
    return out

def fixed_chunks(text, size):
    """Every `size` words, no matter where sentences end."""
    _, words = parse_document(text)
    return [{"text": c["text"], "sec": c["sec"]} for c in chunk_document(words, size, 0)]

def sentence_chunks(text, size):
    """Never cut a sentence: pack whole sentences."""
    return pack(sentences_of(text), size)

def recursive_chunks(text, size):
    """Try the biggest natural unit first: whole section, then sentences, then fixed pieces."""
    sents, out = sentences_of(text), []
    for sec in dict.fromkeys(s["sec"] for s in sents):
        ss = [s for s in sents if s["sec"] == sec]
        n = sum(s["words"] for s in ss)
        if n <= size:
            out.append({"text": " ".join(s["text"] for s in ss), "words": n, "sec": sec})
        else:
            out.extend(pack(ss, size))
    return out

def semantic_chunks(text, size, threshold, lang, stop):
    """Start a new chunk where the topic changes: next sentence too different from the last."""
    sents = sentences_of(text)
    _, _, vectors = embed_chunks(sents, stop, lang)
    out, cur, n = [], [], 0
    for i, s in enumerate(sents):
        sim = cosine(vectors[i - 1], vectors[i]) if i else 1.0
        if cur and (sim < threshold or n + s["words"] > size):
            out.append({"text": " ".join(x["text"] for x in cur), "words": n, "sec": cur[0]["sec"]})
            cur, n = [], 0
        cur.append(s); n += s["words"]
    if cur:
        out.append({"text": " ".join(x["text"] for x in cur), "words": n, "sec": cur[0]["sec"]})
    return out

def quality(chunks):
    """How many chunks, their average size, and how many stop in the middle of a sentence."""
    mid = sum(not re.search(r"[.!?]$", c["text"]) for c in chunks)
    avg = sum(len(c["text"].split()) for c in chunks) / max(1, len(chunks))
    return {"n": len(chunks), "avg": round(avg, 1), "mid_cut": mid}

def demo(text, size, threshold, lang, stop):
    rows = {
        "fixed": fixed_chunks(text, size),
        "sentence": sentence_chunks(text, size),
        "recursive": recursive_chunks(text, size),
        "semantic": semantic_chunks(text, size, threshold, lang, set(stop)),
    }
    summary = {k: quality(v) for k, v in rows.items()}
    print(f"{'chunker':10s} {'chunks':>6s} {'avg words':>9s} {'mid-sentence':>12s}")
    for k, q in summary.items():
        print(f"{k:10s} {q['n']:6d} {q['avg']:9.1f} {q['mid_cut']:12d}")
    return summary
```

```python
# NalarNusa RAG Visual Lab · Lesson 26 · Keeping the index fresh: content hashes, duplicates and four sync strategies
import re

STRATEGIES = ["none", "add-only", "full", "incremental"]

def documents(handbook):
    """One document per section, keyed by a slug of its title."""
    docs = []
    for part in re.split(r"\n(?=## )", handbook.strip()):
        title = part.split("\n", 1)[0].lstrip("# ").strip()
        docs.append({"id": re.sub(r"[^a-z0-9]+", "-", title.lower()).strip("-"), "text": part})
    return docs

def snapshot(docs, day, changes):
    """The folder as it looks on a given day: edits, a re-uploaded copy, an old version, a deletion."""
    out = [dict(d) for d in docs]
    for d in out:
        for when, doc_id, old, new in changes["edits"]:
            if day >= when and d["id"] == doc_id:
                d["text"] = d["text"].replace(old, new)
    for when, doc_id, src, old, new in changes["copies"]:
        if day >= when:
            base = next(d for d in out if d["id"] == src)
            out.append({"id": doc_id, "text": base["text"].replace(old, new)})
    return [d for d in out if not any(day >= when and d["id"] == x for when, x in changes["deletes"])]

def normalize(text):
    return " ".join(text.lower().split())

def fnv1a(text):
    """A 32-bit content hash: same text, same hash, so an unchanged chunk is never embedded twice."""
    h = 0x811C9DC5
    for ch in text:
        h = ((h ^ ord(ch)) * 16777619) & 0xFFFFFFFF
    return format(h, "08x")

def chunk_and_hash(docs, size=50):
    return [{"doc": d["id"], "text": c["text"], "hash": fnv1a(normalize(c["text"]))}
            for d in docs for c in recursive_chunks(d["text"], size)]

def shingles(text):
    w = normalize(text).split()
    return {" ".join(w[i:i + 3]) for i in range(max(1, len(w) - 2))}

def dedup(chunks, threshold):
    """Drop exact copies (same hash) and near copies (word-shingle Jaccard >= threshold)."""
    kept, dropped = [], []
    for c in chunks:
        if any(k["hash"] == c["hash"] for k in kept):
            dropped.append({**c, "why": "exact"})
            continue
        s = shingles(c["text"])
        best = max((len(s & shingles(k["text"])) / len(s | shingles(k["text"])) for k in kept), default=0.0)
        if best >= threshold:
            dropped.append({**c, "why": "near"})
        else:
            kept.append(c)
    return kept, dropped

def wanted(docs, day, changes, threshold):
    kept, _ = dedup(chunk_and_hash(snapshot(docs, day, changes)), threshold)
    return {c["hash"]: c for c in kept}

def sync(docs, day, changes, threshold, strategy):
    """Replay days 1..day. Returns the index on that day and how many chunks were embedded in total."""
    index, embedded, prev = {}, 0, {}
    for t in range(1, day + 1):
        want = wanted(docs, t, changes, threshold)
        if strategy == "none":                    # built once on day 1, never touched again
            if t == 1:
                index, embedded = dict(want), len(want)
        elif strategy == "full":                  # throw it away and re-embed everything
            index, embedded = dict(want), embedded + len(want)
        elif strategy == "add-only":              # embeds what is new, never deletes
            embedded += sum(h not in index for h in want)
            index.update({h: c for h, c in want.items() if h not in index})
        else:                                     # incremental: add new hashes, delete missing ones
            embedded += sum(h not in prev for h in want)
            index = dict(want)
        prev = want
    return list(index.values()), embedded

def ask(index, question, lang, stop, min_sim=0.2):
    idf, _, vectors = embed_chunks(index, stop, lang)
    qv = vectorize(tokenize(question, stop, lang), idf)
    scores = [cosine(qv, v) if qv else 0.0 for v in vectors]
    top = sorted(range(len(scores)), key=lambda i: (-scores[i], i))[:3]
    known = [t for t in tokenize(question, stop, lang) if t in idf]
    used = [(i, scores[i]) for i in top if scores[i] >= min_sim]
    return used, extract_answer(used, index, known, idf, stop, lang)

def verdict(used, index, answer, fact):
    """current, stale, conflict (old and new both in the context), served-deleted, refused or missed."""
    if fact["deleted"]:
        return "refused" if not used else "served-deleted" if fact["new"] in answer else "missed"
    ctx = " ".join(index[i]["text"] for i, _ in used)
    old = fact["old"] if fact["old"] != fact["new"] else None
    if old and old in ctx and fact["new"] in ctx:
        return "conflict"
    if old and old in answer:
        return "stale"
    return "current" if fact["new"] in answer else "missed"

def demo(handbook, changes, facts, day, strategy, threshold, qi, question, lang, stop):
    stop = set(stop)
    docs = documents(handbook)
    today = chunk_and_hash(snapshot(docs, day, changes))
    kept, dropped = dedup(today, threshold)
    index, embedded = sync(docs, day, changes, threshold, strategy)
    cost = {s: sync(docs, day, changes, threshold, s)[1] for s in STRATEGIES}
    used, answer = ask(index, question, lang, stop)
    f = facts[qi]
    fact = {"old": f["old"], "new": f["new"] if day >= f["day"] else f["old"], "deleted": f.get("deleted", False) and day >= f["day"]}
    v = verdict(used, index, answer, fact)
    print(f"day {day}: {len(today)} chunks, {len(dropped)} duplicates dropped; index holds {len(index)}")
    print("embedded so far:", ", ".join(f"{s} {n}" for s, n in cost.items()))
    print(question, "->", answer, f"[{v}]")
    return {"dropped": [c["hash"] for c in dropped], "index": sorted(c["hash"] for c in index), "cost": cost, "answer": answer, "verdict": v}
```

```python
# NalarNusa RAG Visual Lab · Lesson 26 · Incremental sync of a folder into a vector database, with content hashes as ids
# pip install chromadb        (uses Chroma's built-in small embedding model)
import hashlib
import pathlib
import chromadb

FOLDER = pathlib.Path("docs")                 # your documents, as .md or .txt files
client = chromadb.PersistentClient(path="index")
col = client.get_or_create_collection("handbook")


def read_folder(folder):
    return {p.name: p.read_text(encoding="utf-8") for p in sorted(folder.glob("*.*")) if p.suffix in (".md", ".txt")}


def chunks_of(text, size=120):
    words = text.split()
    return [" ".join(words[i:i + size]) for i in range(0, len(words), size)]


def content_id(text):
    """The id IS the hash of the normalised text: same text, same id."""
    return hashlib.sha256(" ".join(text.lower().split()).encode()).hexdigest()[:16]


def sync(folder):
    wanted = {}
    for name, text in read_folder(folder).items():
        for piece in chunks_of(text):
            wanted.setdefault(content_id(piece), (piece, name))   # an exact copy has the same id: kept once
    existing = set(col.get(include=[])["ids"])
    new = [h for h in wanted if h not in existing]
    gone = [h for h in existing if h not in wanted]
    if new:
        col.add(ids=new, documents=[wanted[h][0] for h in new], metadatas=[{"source": wanted[h][1]} for h in new])
    if gone:
        col.delete(ids=gone)                                       # deletions matter as much as additions
    print(f"{len(new)} embedded, {len(gone)} deleted, {len(wanted) - len(new)} unchanged")


if __name__ == "__main__":
    FOLDER.mkdir(exist_ok=True)
    sync(FOLDER)                                                   # run it again after editing files: only changes are embedded
    res = col.query(query_texts=["When is the lab open on Saturday?"], n_results=3)
    for doc, meta in zip(res["documents"][0], res["metadatas"][0]):
        print(meta["source"], "|", doc[:100])
```
