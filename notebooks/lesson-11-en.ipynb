## Part 1 · From scratch

```python
# NalarNusa RAG Visual Lab · Lesson 11 · Exact search vs a graph index (NSW, the idea behind HNSW)

def rng(seed):
    s = seed % 2147483646 + 1
    def rand():
        nonlocal s
        s = s * 16807 % 2147483647
        return s / 2147483647
    return rand

def make_points(n, seed, clusters=6):
    r = rng(seed)
    centers = [(0.12 + r() * 0.76, 0.12 + r() * 0.76) for _ in range(clusters)]
    pts = []
    for i in range(n):
        cx, cy = centers[i % clusters]
        jx = (r() + r() + r()) / 3 - 0.5
        jy = (r() + r() + r()) / 3 - 0.5
        pts.append((min(1, max(0, cx + jx * 0.4)), min(1, max(0, cy + jy * 0.4))))
    return pts

def d2(a, b):
    return (a[0] - b[0]) ** 2 + (a[1] - b[1]) ** 2

def build_nsw(points, M):
    """Insert points one by one; link each to its M nearest earlier points, both ways."""
    nbrs = [[] for _ in points]
    for i in range(1, len(points)):
        cand = sorted((d2(points[i], points[j]), j) for j in range(i))
        for _, j in cand[:M]:
            nbrs[i].append(j)
            nbrs[j].append(i)
    return nbrs

def brute_force(points, q, k):
    """Compare the query with every single vector: exact, but cost grows with the collection."""
    return [i for _, i in sorted((d2(p, q), i) for i, p in enumerate(points))[:k]], len(points)

def search_nsw(points, nbrs, q, k, ef, entry=0):
    """Walk the graph: always expand the closest unexplored point, stop when nothing closer is left."""
    count = 0
    def dist(i):
        nonlocal count
        count += 1
        return d2(points[i], q)
    visited = {entry}
    d0 = dist(entry)
    cand, res = [(d0, entry)], [(d0, entry)]
    width = max(ef, k)
    while cand:
        cand.sort()
        d, c = cand.pop(0)
        if len(res) >= width and d > res[-1][0]:
            break
        for nb in nbrs[c]:
            if nb in visited:
                continue
            visited.add(nb)
            dn = dist(nb)
            if len(res) < width or dn < res[-1][0]:
                cand.append((dn, nb))
                res.append((dn, nb))
                res.sort()
                if len(res) > width:
                    res.pop()
    return [i for _, i in res[:k]], count

def run(n, seed, M, ef, qx, qy, k=5):
    pts = make_points(n, seed)
    nbrs = build_nsw(pts, M)
    exact, c1 = brute_force(pts, (qx, qy), k)
    found, c2 = search_nsw(pts, nbrs, (qx, qy), k, ef)
    recall = len(set(exact) & set(found)) / k
    print(f"brute force: {c1} comparisons; graph: {c2} comparisons; recall@{k} = {recall:.0%}")
    return {"exact": exact, "found": found, "brute": c1, "graph": c2}
```

```python
# NalarNusa RAG Visual Lab · Lesson 11 · Approximate nearest-neighbour search with FAISS (HNSW)
# pip install faiss-cpu numpy
import time
import faiss
import numpy as np

rng = np.random.default_rng(7)
centres = rng.normal(size=(200, 384)).astype(np.float32)          # real embeddings form topic clusters
X = centres[rng.integers(0, 200, 100_000)] + 0.35 * rng.normal(size=(100_000, 384)).astype(np.float32)
faiss.normalize_L2(X)                                             # 100k vectors, embedding-sized, length 1
Q = X[:20] + 0.05 * rng.normal(size=(20, 384)).astype(np.float32)
faiss.normalize_L2(Q)                                             # 20 queries

hnsw = faiss.IndexHNSWFlat(384, 16)         # M = 16 links per vector; L2 on length-1 vectors ranks like cosine
hnsw.add(X)                                 # the graph is built once, here

flat = faiss.IndexFlatL2(384); flat.add(X)  # exact: compares the query with every vector
t = time.perf_counter(); _, exact = flat.search(Q, 5); t_flat = (time.perf_counter() - t) / len(Q)

for ef in (16, 32, 128):                    # search width (ef): wider = slower, finds more
    hnsw.hnsw.efSearch = ef
    t = time.perf_counter(); _, found = hnsw.search(Q, 5); t_hnsw = (time.perf_counter() - t) / len(Q)
    recall = np.mean([len(set(a) & set(b)) / 5 for a, b in zip(found, exact)])
    print(f"ef {ef:3d}: recall@5 {recall:.0%}   graph {t_hnsw * 1e3:.2f} ms vs exact {t_flat * 1e3:.2f} ms per query")
```
